## Importing the Required Libraries

```python
import fitz
import re
```

## Creating Functions for reusibility of the code

```python
def extract_text_from_pdf(pdf_path):
    """Extracts raw text from the PDF."""
    doc = fitz.open(pdf_path)
    text = ""
    for page in doc:
        text += page.get_text("text") + "\n"
    return text

def split_into_clauses(text):
    """
    Splits text based on standard legal numbering and headers.
    Regex matches: "1.", "1.1", "Section 2", "Clause 3(a)"
    """
    # This regex looks for start-of-line clause indicators
    pattern = re.compile(r'(?i)^(?:Section\s+\d+|Clause\s+\d+|\d+\.\d+|\d+\.)\s*', re.MULTILINE)
    
    splits = list(pattern.finditer(text))
    clauses = []
    
    # Slice the text between regex matches
    for i, match in enumerate(splits):
        start = match.start()
        # End is the start of the next match, or the end of the document
        end = splits[i+1].start() if i + 1 < len(splits) else len(text)
        
        clause_text = text[start:end].strip()
        
        # Clean up excessive newlines
        clause_text = re.sub(r'\n+', ' ', clause_text)
        clauses.append(clause_text)
        
    return clauses
```

## TESTING IT OUT

```python
# Replace with the path to a downloaded sample T&C PDF
pdf_text = extract_text_from_pdf("../data/raw/ABEMA.pdf") 

clauses = split_into_clauses(pdf_text)

# Print the first 10 chunks to verify the splitting
for idx, clause in enumerate(clauses[:10]):
    print(f"--- Chunk {idx + 1} ---")
    print(clause)
    print()
```
